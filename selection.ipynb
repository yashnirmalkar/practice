```python
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Rohit"],
        "Age": [20, 21, 19, 22, 20],
        "Marks": [85, 92, 78, 88, 95],
        "City": ["Raipur", "Delhi", "Mumbai", "Raipur", "Delhi"]
    
}

df = pd.DataFrame(data)

def assign_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"


# Apply the function to create the Grade column
df["Grade"] = df["Marks"].apply(assign_grade)

print(df)
```

```python
# Sort students by Marks in descending order
sorted_df = df.sort_values(by="Marks", ascending=False)

print(sorted_df)
```

```python
#find age greater 20
#syntax - np.where(df[name]condtition , if true , if false)
import numpy as np
df['greater age ']=np.where(df['Age']>=20, "eligible", "not eligible")
print(df)
```

```python
delhi_students = df[df["City"] == "Delhi"]

print(delhi_students)
```

```python
#give bonus marks
import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Rohit"],
    "Age": [20, 21, 19, 22, 20],
    "Marks": [85, 92, 78, 88, 95],
    "City": ["Raipur", "Delhi", "Mumbai", "Raipur", "Delhi"]
}

df = pd.DataFrame(data)

df['bonus']=5
print(df)
```

```python
#create final marks
df['finalMarks']=df['Marks']+df['bonus']
print(df)
```

```python
df['City']=df['City'].replace('Mumbai', 'Pune')
print(df)
```

```python
df = df.rename(columns={"Marks": "Original_Marks"})
print(df)
```

```python
del df['bonus']
print(df)
```
